### Task 13: Implement Adaptive Median Filter
Write a custom function for adaptive median filtering. It should dynamically increase window size until noise is removed or max size is reached. Apply to a noisy image and compare with standard median.

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

def add_salt_pepper_noise(image, density=0.05):
    
    if image.ndim == 3:
        gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
    else:
        gray = image.copy()

    rows, cols = gray.shape
    noisy = gray.copy()
    num_salt = int(np.ceil(density * rows * cols / 2))
    num_pepper = int(np.ceil(density * rows * cols / 2))
    rs = np.random.randint(0, rows, num_salt)
    cs = np.random.randint(0, cols, num_salt)
    noisy[rs, cs] = 255

    rp = np.random.randint(0, rows, num_pepper)
    cp = np.random.randint(0, cols, num_pepper)
    noisy[rp, cp] = 0

    return noisy

def adaptive_median_filter(image, max_size=7):
 
    if image.ndim == 3:
        gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
    else:
        gray = image.copy()

    if max_size % 2 == 0:
        max_size += 1

    pad = max_size // 2
    padded = np.pad(gray, pad, mode='edge')
    rows, cols = gray.shape
    out = np.zeros_like(gray, dtype=np.uint8)

   
    for i in range(rows):
        for j in range(cols):
            window_size = 3
            while True:
                half = window_size // 2
                r1 = i + pad - half
                r2 = i + pad + half + 1
                c1 = j + pad - half
                c2 = j + pad + half + 1

                window = padded[r1:r2, c1:c2]

                z_min = int(window.min())
                z_max = int(window.max())
                z_med = int(np.median(window))
                z_xy = int(padded[i + pad, j + pad])

                A1 = z_med - z_min
                A2 = z_med - z_max

                # Stage A
                if A1 > 0 and A2 < 0:
                    # Stage B
                    B1 = z_xy - z_min
                    B2 = z_xy - z_max
                    if B1 > 0 and B2 < 0:
                        out[i, j] = z_xy
                    else:
                        out[i, j] = z_med
                    break
                else:
                    window_size += 2
                    if window_size > max_size:
                        out[i, j] = z_med
                        break

    return out


if __name__ == "__main__":
    
    img_path = "noisy_image.jpg"   
    img = cv.imread(img_path, cv.IMREAD_GRAYSCALE)
    if img is None:
        print(f"Warning: '{img_path}' not found — using synthetic test image.")
        
        img = np.tile(np.linspace(30, 220, 256, dtype=np.uint8), (256, 1))

    
    noisy = add_salt_pepper_noise(img, density=0.07)

    
    med3 = cv.medianBlur(noisy, 3)
    med5 = cv.medianBlur(noisy, 5)

    
    adaptive = adaptive_median_filter(noisy, max_size=7)

   
    fig, axes = plt.subplots(2, 4, figsize=(16, 8))
    axes = axes.ravel()

    axes[0].imshow(img, cmap='gray', vmin=0, vmax=255)
    axes[0].set_title("Original (grayscale)")
    axes[0].axis('off')

    axes[1].imshow(noisy, cmap='gray', vmin=0, vmax=255)
    axes[1].set_title("Noisy (salt & pepper)")
    axes[1].axis('off')

    axes[2].imshow(med3, cmap='gray', vmin=0, vmax=255)
    axes[2].set_title("Median (3x3)")
    axes[2].axis('off')

    axes[3].imshow(med5, cmap='gray', vmin=0, vmax=255)
    axes[3].set_title("Median (5x5)")
    axes[3].axis('off')

    axes[4].imshow(adaptive, cmap='gray', vmin=0, vmax=255)
    axes[4].set_title("Adaptive Median (max 7x7)")
    axes[4].axis('off')

    
    r0, c0, w = img.shape[0]//2 - 40, img.shape[1]//2 - 40, 80
    axes[5].imshow(img[r0:r0+w, c0:c0+w], cmap='gray', vmin=0, vmax=255)
    axes[5].set_title("Original (zoom)")
    axes[5].axis('off')

    axes[6].imshow(noisy[r0:r0+w, c0:c0+w], cmap='gray', vmin=0, vmax=255)
    axes[6].set_title("Noisy (zoom)")
    axes[6].axis('off')

    axes[7].imshow(adaptive[r0:r0+w, c0:c0+w], cmap='gray', vmin=0, vmax=255)
    axes[7].set_title("Adaptive (zoom)")
    axes[7].axis('off')

    plt.tight_layout()
    plt.show()
```

### Task 14: Implement Bilateral Filter Function
Write a Python function to perform bilateral filtering on an image. Use Gaussian weights for both spatial and intensity. Parameters: diameter, sigma_color, sigma_space. Compare with cv.bilateralFilter().

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

def custom_bilateral_filter(img, diameter, sigma_color, sigma_space):
    pass
    if len(img.shape) == 3:
        img = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

    rows, cols = img.shape
    half = diameter // 2
    padded = np.pad(img, ((half, half), (half, half)), mode="reflect")
    filtered = np.zeros_like(img, dtype=np.float32)

    x, y = np.mgrid[-half:half+1, -half:half+1]
    spatial_gauss = np.exp(-(x**2 + y**2) / (2 * sigma_space**2))

    
    for i in range(rows):
        for j in range(cols):
            region = padded[i:i+diameter, j:j+diameter]
            center_val = img[i, j]
            diff = region - center_val
            range_gauss = np.exp(-(diff**2) / (2 * sigma_color**2))

            weights = spatial_gauss * range_gauss
            weights /= np.sum(weights)
            filtered[i, j] = np.sum(region * weights)

    return filtered.astype(np.uint8)



img = cv.imread(r"c:/Users/maryam/Desktop/task22/SkyXperts-Vision-Course/Session1/images/dog.jpg")
img_gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
img_gray = cv.resize(img_gray, (200, 200))  

custom_result = custom_bilateral_filter(img_gray, diameter=5, sigma_color=25, sigma_space=25)

opencv_result = cv.bilateralFilter(img_gray, 5, 25, 25)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.title("Original (Gray)")
plt.imshow(img_gray, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Custom Bilateral")
plt.imshow(custom_result, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("OpenCV Bilateral")
plt.imshow(opencv_result, cmap="gray")
plt.axis("off")

plt.show()
```
